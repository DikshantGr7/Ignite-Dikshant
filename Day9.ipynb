```python
x = np.array(['Delhi','Mumbai','Chennai','Kolkata'])
y = np.array([35,67,22,43])
plt.barh(x, y, color="green", height=0.5)
plt.title("Thermal Power Plants in Cities", color="Black", loc="left")
plt.show()
```

```python
x = np.random.normal(150,10,250)
plt.hist(x, linestyle = "--",edgecolor = "Black",rwidth= 0.9,alpha = 0.8,bins = 9)
plt.show()
```

```python
y = np.array([19,12,23,11])
my_label = ["Delhi","Kolkata","Chennai","Mumbai"]
my_explode =[0.2,0,0,0]
my_color = ["Blue","Green","Red","Yellow"]
plt.title("Nuclear Facilities in Metropolitan Cities",loc = "left")
plt.pie(y,shadow = True,explode = my_explode, colors = my_color,labels = my_label)
plt.legend(title = "City Name")
plt.show()
```

```python
import random

colors = np.random.randint(100,size = (5))
sizes = 10 * np.random.randint(100,size = (5))

x1 = np.array([7,9,10,25,78])
y1 = np.array([23,44,11,10,19])
plt.scatter(x1,y1,cmap = "nipy_spectral",c = colors,s = sizes)

x2 = np.array([12,23,45,19,10])
y2 = np.array([16,17,23,19,24])
plt.scatter(x2,y2,cmap = "nipy_spectral",c = colors,s = sizes)

plt.show()
```

```python
import matplotlib.pyplot as plt
import pandas as pd
import seaborn as sns

df = pd.read_csv("orders.csv")

plt.figure(figsize=(4, 12))

sns.heatmap(
    df[["Price"]],
    annot=True,
    fmt=".2f",
    cmap="YlGnBu",
    cbar=True,
    linewidths=0.5,  
    linecolor="white",
    annot_kws={"size": 9},  
    xticklabels=False,  
)

plt.title("Price Heatmap", pad=15)
plt.ylabel("Index / Row")
plt.tight_layout()
plt.show()
```
