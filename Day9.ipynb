```python
import matplotlib.pyplot as plt
import numpy as np
import pandas as pd
```

```python
x_points = np.array([25,13])
y_points = np.array([11,19])
plt.plot(x_points,y_points)
plt.show()
```

```python
x_points = np.array([25,13])
y_points = np.array([11,19])
plt.plot(x_points,y_points,"o")
plt.show()
```

```python
x_points = np.array([25,13,12,0])
y_points = np.array([11,19,13,11])
plt.plot(x_points,y_points)
plt.show()
```

```python
y_points = np.array([11,19])
plt.plot(y_points, marker = "o")
plt.show()
```

```python
x1 = np.array([25,13])
y1 = np.array([11,19])
x2 = np.array([4,13])
y2 = np.array([13,14])
plt.plot(x1,y1,marker = "o",ms = 20,mfc = "r", mec = 'black',ls = ":", color = "red", linewidth = "5")
plt.plot(x2,y2,marker = "o",ms = 20,mfc = "g", mec = 'black',ls = ":", color = "hotpink", linewidth = "5")
plt.show()
```
